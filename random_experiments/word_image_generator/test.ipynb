```python
from transformers import VisionEncoderDecoderModel, TrOCRProcessor
from PIL import Image
```

```python
image = Image.open("/Users/xai/Personal/Projects/TeluguOCR/image_encoder/words/images/image_26.png")

processor = TrOCRProcessor.from_pretrained('microsoft/trocr-small-printed')
model = VisionEncoderDecoderModel.from_pretrained('microsoft/trocr-small-printed')
```

```python
print(model)
```

```python
encoder_model = model.encoder

processed_image = processor(images=image, return_tensors="pt").pixel_values
print(processed_image.size())
```

```python
output = encoder_model(processed_image)
output.last_hidden_state.shape
```

```python
temp_output = encoder_model.embeddings.patch_embeddings(processed_image)
print(temp_output.shape)
temp_output = model(processed_image)
print(temp_output.last_hidden_state.shape)
```
